- Returns `a / b` normally
- Returns `None` if `b == 0` (catch `ZeroDivisionError`)
- Raises `TypeError` with message `"Inputs must be numeric"` if either input is not a number

````python
def safe_divide(a, b):
    if not (isinstance(a, (int, float)) and isinstance(b, (int, float))):
        raise TypeError("Inputs must be numeric")
    if b == 0:
        return None
    return a / b

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

### 4.3 Eigenvalues & Eigenvectors

For the matrix:
```

     [2, 3]]
```

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.
2. **Verify** the eigenvector equation: `A @ v = λ * v` for each eigenpair.
3. Plot both eigenvectors as arrows, scaled by their eigenvalues, on a 2D plot.
4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v   = eigenvectors[:, i]
    assert np.allclose(A @ v, lam * v), f"Eigenpair {i} failed verification"
    print(f"λ={lam:.1f}: A@v = {A@v}, λ*v = {lam*v}  ✓")

# Plot
fig, ax = plt.subplots()
origin = np.zeros(2)
colors = ['steelblue', 'tomato']
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    ax.quiver(*origin, *v, angles='xy', scale_units='xy', scale=1,
              color=colors[i], width=0.02, label=f'λ={eigenvalues[i]:.1f}')
ax.set_xlim(-1.5, 1.5); ax.set_ylim(-1.5, 1.5)
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.set_title('Eigenvectors of A')
ax.legend(); plt.tight_layout(); plt.show()

print("4.3 passed ✓")
````

when you multiply a vector by a matrix it usually rotates AND scales it. but eigenvectors are special — the matrix only stretches or squishes them, no rotation. the eigenvalue λ is just how much that stretching happens, so λ=5 means 5x longer, λ<0 means it flips direction, λ=0 means it collapses to zero. in PCA we literally use this to find the directions where data stretches the most, which turn out to be the directions carrying the most information.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=True)

X_reconstructed = U[:, :3] @ np.diag(S) @ Vt
X_approx        = S[0] * np.outer(U[:, 0], Vt[0, :])

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

SVD splits any matrix X into three pieces — U, S, and Vt. let me break down what each one actually is.
Vt (or V transposed) is the one that matters most for PCA. the rows of Vt are your principal components — literally the directions of maximum variance in your data. so if you want to know "which direction does my data spread out the most", thats the first row of Vt. second most variance? second row. and so on.
S is the diagonal matrix of singular values. these tell you HOW MUCH variance exists along each of those Vt directions. bigger singular value = more variance in that direction. you can actually square the singular values to get the eigenvalues of the covariance matrix, which is kinda neat.
U is the left singular vectors — basically the coordinates of your original data points after projecting them onto those principal component directions. so U @ diag(S) gives you the actual transformed data in PCA space.
the connection to PCA is pretty direct tbh. if you center your data first (subtract the mean), then running SVD on X gives you basically the same result as computing eigenvectors of the covariance matrix X.T @ X — just way more numerically stable. thats actually why most PCA implementations under the hood just use SVD instead of explicitly computing the covariance matrix.
so yeah — Vt has the directions, S has the importance of each direction, and U has where your data points land after the transformation.

**H₀ / H₁ and Interpretation (5.2):**

H₁: the alternative hypothesis — what you actually think might be true. here its saying engineering salaries ARE different from the overall mean. this is what you're rooting for if you think theres a real effect.
Conclusion: so you look at the p-value and if its below 0.05 you reject H₀ and say "ok something real is going on here". if its above 0.05 you fail to reject H₀ — and note that "fail to reject" is NOT the same as saying H₀ is true, it just means you dont have enough evidence to confidently say otherwise.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm')
ks_s2 = stats.kstest(s2, 'norm')
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff  = np.diff(ts)
adf_diff = adfuller(ts_diff)
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

### 5.5 Model Monitoring Concepts

1. Implement `compute_psi(expected, actual, bins=10)` that computes the **Population Stability Index** between two distributions.
2. Simulate a scenario where a model's input distribution shifts and plot both distributions.
3. In the markdown cell, define: *concept drift*, *covariate drift*, PSI thresholds (what values indicate no/minor/major shift), and one trigger for retraining.

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    eps = 1e-10
    breakpoints = np.percentile(expected, np.linspace(0, 100, bins + 1))

    expected_counts = np.histogram(expected, bins=breakpoints)[0]
    actual_counts   = np.histogram(actual,   bins=breakpoints)[0]

    expected_pct = expected_counts / len(expected) + eps
    actual_pct   = actual_counts   / len(actual)   + eps

    psi = np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))
    return psi

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)
drift_dist = np.random.normal(65, 12, 1000)

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
ax.hist(train_dist, bins=30, alpha=0.6, color='steelblue', density=True, label='Training')
ax.hist(drift_dist, bins=30, alpha=0.6, color='tomato',    density=True, label='Production')
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

Concept drift is when the relationship between your input features and the target variable changes over time. so like if i trained a loan default model before covid, the model didnt suddenly become bad code — the whole pattern of who defaults just changed in the real world. the X→Y mapping itself shifted underneath me.
Covariate drift is different — here the input distribution changes but the relationship between inputs and target stays the same. so if i trained my model mostly on young users and suddenly older users start using the product, the underlying patterns are fine, just a different crowd showing up in my data.
for PSI thresholds — if PSI is below 0.1 the distribution is stable and i dont need to do anything. if its between 0.1 and 0.2 theres a minor shift happening and i'd start monitoring more closely but its not urgent. if it crosses 0.2 thats my red flag — the shift is significant enough that my model is probably already degrading and i need to retrain.
a practical retraining trigger would be something like — if im running a fraud detection model and my PSI on transaction amount crosses 0.25, that means the spending patterns coming in look nothing like what i trained on. at that point i pull fresh labeled data and retrain before my false negative rate gets out of hand.

### 6.1 Core Concepts

A bag contains: 4 red, 3 blue, 3 green marbles.

1. Define the sample space and compute P(red), P(blue), P(green).
2. Two marbles drawn **without replacement**. Compute:
   - P(first=red, second=blue) — **joint probability**
   - P(second=blue | first=red) — **conditional probability**
3. Are the two draws **independent**? Show mathematically.

````python
total = 10
red, blue, green = 4, 3, 3

p_red   = red / total
p_blue  = blue / total
p_green = green / total

# Joint: P(red first AND blue second) — without replacement
p_red_then_blue = (red / total) * (blue / (total - 1))

# Conditional: P(blue second | red first)
p_blue_given_red = blue / (total - 1)

# Independence check: P(B|A) == P(B)?
independent = np.isclose(p_blue_given_red, p_blue)

print(f"P(red)={p_red:.3f}, P(blue)={p_blue:.3f}, P(green)={p_green:.3f}")
print(f"P(red,blue): {p_red_then_blue:.4f}")
print(f"P(blue|red): {p_blue_given_red:.4f}")
print(f"P(blue):     {p_blue:.4f}")
print(f"Independent? {independent}")
````

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    prior_ham = 1 - prior_spam
    p_word = (p_word_given_spam * prior_spam) + (p_word_given_ham * prior_ham)
    return (p_word_given_spam * prior_spam) / p_word

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30
sample_means = np.array([np.random.choice(population, size=sample_size, replace=False).mean()
                         for _ in range(n_samples)])

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
ax.hist(sample_means, bins=50, density=True, color='steelblue', alpha=0.7, label='Sample Means')
x = np.linspace(sample_means.min(), sample_means.max(), 300)
ax.plot(x, stats.norm.pdf(x, pop_mean, clt_std), color='tomato', lw=2, label='Theoretical Normal')
ax.legend()
ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

the CLT matters because no matter what weird shape your population distribution has — exponential, skewed, whatever — if you take enough samples the means will always form a normal distribution. this is huge because basically every classical statistical test (t-tests, z-tests, confidence intervals) assumes normality, and the CLT is the reason we can get away with that assumption even when the underlying data isnt normal at all. in ML its also why things like gradient estimates in SGD behave predictably — youre averaging over mini-batches and that averaging effect kicks in.
